# Question 3: Perceptron Learning Algorithm

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.array([
    [ 1.0,  1.0,  1],
    [-1.0, -1.0, -1],
    [ 0.0,  0.5, -1],
    [ 0.1,  0.5, -1],
    [ 0.2,  0.2,  1],
    [ 0.9,  0.5,  1],
])

x1 = data[:, 0]
x2 = data[:, 1]
y  = data[:, 2].astype(int)

# Scatter plot coloured by class label
colors = {-1: 'steelblue', 1: 'tomato'}
labels = {-1: 'y = -1', 1: 'y = 1'}

fig, ax = plt.subplots()
for cls in (-1, 1):
    mask = y == cls
    ax.scatter(x1[mask], x2[mask], c=colors[cls], label=labels[cls], s=80, edgecolors='k')

ax.set_xlabel('x₁')
ax.set_ylabel('x₂')
ax.set_title('Scatter plot of x₁ vs x₂')
ax.legend()
ax.grid(True, linestyle='--', linewidth=0.6, alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
def plot_with_boundary(x1, x2, y, w, b, step_label, colors, labels):
    w1, w2 = w
    fig, ax = plt.subplots()
    for cls in (-1, 1):
        mask = y == cls
        ax.scatter(x1[mask], x2[mask], c=colors[cls], label=labels[cls], s=80, edgecolors='k')

    # Decision boundary: w1*x1 + w2*x2 + b = 0  =>  x2 = -(w1*x1 + b) / w2
    x1_range = np.linspace(x1.min() - 0.2, x1.max() + 0.2, 200)
    if abs(w2) > 1e-9:
        x2_boundary = -(w1 * x1_range + b) / w2
        ax.plot(x1_range, x2_boundary, 'k-', linewidth=1.5, label='Decision boundary')
    else:
        if abs(w1) > 1e-9:
            xv = -b / w1
            ax.axvline(xv, color='k', linewidth=1.5, label='Decision boundary')

    ax.set_xlabel('x₁')
    ax.set_ylabel('x₂')
    ax.set_title(f'Decision boundary — {step_label}')
    ax.legend()
    ax.grid(True, linestyle='--', linewidth=0.6, alpha=0.7)
    plt.tight_layout()
    plt.show()


# ── Perceptron training ──────────────────────────────────────────────────────
X = np.column_stack([x1, x2])   # no bias column — bias handled separately
labels_plot = {-1: 'y = -1', 1: 'y = 1'}
colors_plot  = {-1: 'steelblue', 1: 'tomato'}

lr = 1.0           # simplified learning rate
w  = np.ones (2)   # [w1, w2]
b  = 0.0           # bias
max_epochs = 20

print("=" * 60)
print(f"Initial weights:  w1={w[0]:.4f}, w2={w[1]:.4f},  b={b:.4f}")
print("=" * 60)

update_count = 0
converged    = False

for epoch in range(1, max_epochs + 1):
    errors_this_epoch = 0
    for i, (xi, yi) in enumerate(zip(X, y)):
        net   = np.dot(w, xi) + b
        y_hat = int(np.sign(net)) if net != 0 else -1   # sign; treat 0 as -1 (error)

        print(f"\nEpoch {epoch}, Sample {i+1}:  x1={xi[0]:.1f}, x2={xi[1]:.1f}, target={yi}")
        print(f"  Current weights : w1={w[0]:.4f}, w2={w[1]:.4f},  b={b:.4f}")
        print(f"  net = w·x + b = {net:.4f}  →  sign(net) = {y_hat}  (target={yi})")
        print(f"  y × (w·x + b) = {yi} × {net:.4f} = {yi * net:.4f}")

        if yi * net <= 0:
            w = w + lr * yi * xi
            b = b + lr * yi
            update_count += 1
            errors_this_epoch += 1
            print(f"  ✗ Update  (y·(w·x+b) ≤ 0)  →  w1={w[0]:.4f}, w2={w[1]:.4f},  b={b:.4f}")

            # Decision boundary equation
            if abs(w[1]) > 1e-9:
                slope     = -w[0] / w[1]
                intercept = -b    / w[1]
                print(f"\n  Decision boundary: {w[0]:.4f}·x₁ + {w[1]:.4f}·x₂ + {b:.4f} = 0")
                print(f"  ↳  x₂ = {slope:.4f}·x₁ + ({intercept:.4f})")
            else:
                print(f"\n  Decision boundary: {w[0]:.4f}·x₁ + {b:.4f} = 0")
                if abs(w[0]) > 1e-9:
                    print(f"  ↳  x₁ = {(-b/w[0]):.4f}   (vertical line)")

            plot_with_boundary(x1, x2, y, w, b,
                               f"Epoch {epoch}, update #{update_count} after sample {i+1}",
                               colors_plot, labels_plot)
        else:
            print(f"  ✓ No update  (y·(w·x+b) > 0)")

    if errors_this_epoch == 0:
        print(f"\n{'='*60}")
        print(f"✔ Converged after epoch {epoch}  (total weight updates: {update_count})")
        print(f"Final weights: w1={w[0]:.4f}, w2={w[1]:.4f},  b={b:.4f}")
        converged = True
        break

if not converged:
    print(f"\n{'='*60}")
    print(f"Did not converge within {max_epochs} epochs.")
    print(f"Final weights: w1={w[0]:.4f}, w2={w[1]:.4f},  b={b:.4f}")

# Final boundary plot
print("\nFinal decision boundary:")
if abs(w[1]) > 1e-9:
    print(f"  {w[0]:.4f}·x₁ + {w[1]:.4f}·x₂ + {b:.4f} = 0")
    print(f"  ↳  x₂ = {-w[0]/w[1]:.4f}·x₁ + ({-b/w[1]:.4f})")
plot_with_boundary(x1, x2, y, w, b, "Final", colors_plot, labels_plot)
```
